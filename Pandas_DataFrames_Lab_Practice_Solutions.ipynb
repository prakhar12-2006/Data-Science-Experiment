# Section C — Hard (Q26–Q30)

### Q26. Create course_df and perform all requested operations.

```python
data = {'Student': ['Aman', 'Ravi', 'Neha', 'Priya', 'Karan'], 'Course': ['Python', 'Java', 'C++', 'Python', 'Java'], 'Score': [85, 90, 78, 88, 76]}
course_df = pd.DataFrame(data)
print(course_df.columns)
print(course_df['Student'])
new_df = course_df[['Student', 'Score']]
print(new_df)
course_df['Batch'] = '2026'
print(course_df)
```

### Q27. Create travel_df, add Code with np.arange(5), then replace it with a partially indexed Series.

```python
travel_df = pd.DataFrame({'Destination': ['Delhi', 'Agra', 'Jaipur', 'Mumbai', 'Goa']})
travel_df['Code'] = np.arange(5)
print("After adding Code:")
print(travel_df)
codes = pd.Series([101, 104], index=[1, 4])
travel_df['Code'] = codes
print("After replacing Code:")
print(travel_df)
```

### Q28. Create school_df and add a missing House column.

```python
data = {'Student': ['Aman', 'Ravi', 'Neha'], 'Class': [10, 10, 9], 'Section': ['A', 'B', 'A']}
school_df = pd.DataFrame(data)
new_df = pd.DataFrame(school_df, columns=['Student', 'Class', 'Section', 'House'])
new_df['House'] = 'Red House'
print(new_df)
```

### Q29. Create inventory_df, add a partially indexed Series, then delete the added column.

```python
inventory_df = pd.DataFrame({'Item': ['Pen', 'Book', 'Bag', 'Pencil'], 'Quantity': [10, 20, 5, 15]})
stock = pd.Series([100, 300], index=[0, 2])
inventory_df['Stock'] = stock
print("After adding Stock:")
print(inventory_df)
del inventory_df['Stock']
print("After deleting Stock:")
print(inventory_df)
```
