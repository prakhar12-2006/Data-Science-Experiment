# Topic 7: Functions & Passing Arguments

### Q7.1. Write my_abs() without built-in abs().

```python
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

print(my_abs(5))
print(my_abs(-7))
print(my_abs(0))
```

### Q7.2. What happens when my_abs('Hi') is called?

### Q7.3. Write print_abs() and inspect its return value.

```python
def print_abs(val):
    if val < 0:
        print(0 - val)
    else:
        print(val)

x = print_abs(-2.7)
print(x)  # None, because the function has no return statement
```

### Q7.4. Why does inc_val() not change x?

```python
def inc_val(val):
    val = val + 1

x = 7
inc_val(x)
print(x)  # 7
```

### Q7.5. Swap local variables and show originals are unchanged.

```python
def swap_message(val1, val2):
    val1, val2 = val2, val1
    print("Inside function:", val1, val2)

x = 6
y = 3
swap_message(6, 3)
print("Original variables:", x, y)
```

# Topic 8: Scope — Local vs Global Variables

### Q8.1. Explain why print(val) outside the function causes an error.

```python
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

# Uncomment to see NameError: val is local to my_abs().
# print(val)
```

### Q8.2. Read a global variable inside a function.

```python
counter = 0

def show_counter():
    print(counter)

show_counter()
```

### Q8.3. Try to increment counter without global.

```python
counter = 0

def increment_counter():
    # This raises UnboundLocalError if called, because counter is treated
    # as local and is read before it has a local value.
    counter = counter + 1

# Uncomment to observe the error:
# increment_counter()
```

### Q8.4. Increment the global counter correctly.

```python
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
```
