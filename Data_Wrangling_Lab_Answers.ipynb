# Data Science and Visualization Lab
## Data Wrangling Practice Questions: Answers

```python
import numpy as np
import pandas as pd

students = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": ["Asha", "Ravi", "Neha", "Imran", "Kiran", "Meera", "Arjun", "Sara"],
    "Department": ["CSE", "DS", "CSE", "DS", "IT", "IT", "CSE", "DS"]
})
marks = pd.DataFrame({
    "Student_ID": [101, 102, 104, 105, 106, 109],
    "Data_Science_Marks": [86, 91, 78, 64, 72, 88]
})
attendance = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Attendance_Percent": [92, 78, 85, 96, 70, 88, 81, 90]
})
results = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Final_Marks": [88, 73, 81, 94, 60, 79, 76, 91]
})
dept_marks = pd.DataFrame({
    "Department": ["CSE", "CSE", "CSE", "DS", "DS", "DS", "IT", "IT", "IT"],
    "Semester": [1, 2, 3, 1, 2, 3, 1, 2, 3],
    "Average_Marks": [76, 81, 79, 82, 85, 88, 70, 74, 72]
})
marks_by_sem = pd.DataFrame({
    "Semester_1": [76, 82, 70],
    "Semester_2": [81, 85, 74],
    "Semester_3": [79, 88, 72]
}, index=["CSE", "DS", "IT"])
jan = pd.DataFrame({"Month": ["Jan"] * 3, "Product": ["Pen", "Notebook", "Pencil"], "Units": [120, 80, 60]})
feb = pd.DataFrame({"Month": ["Feb"] * 3, "Product": ["Pen", "Notebook", "Pencil"], "Units": [150, 95, 70]})
mar = pd.DataFrame({"Month": ["Mar"] * 3, "Product": ["Pen", "Notebook", "Pencil"], "Units": [130, 110, 90]})
transactions = pd.DataFrame({
    "Product": ["Pen", "Pen", "Pen", "Notebook", "Notebook", "Notebook", "Pencil"],
    "Month": ["Jan", "Jan", "Feb", "Jan", "Feb", "Feb", "Jan"],
    "Units": [50, 70, 150, 80, 70, 25, 60]
})
quiz_wide = pd.DataFrame({
    "Student": ["Asha", "Ravi", "Neha"],
    "Quiz_1": [8, 7, 9],
    "Quiz_2": [9, 8, 10],
    "Quiz_3": [7, 9, 9]
})
```

## Q1. Look before you combine

```python
print(students.head()); print(students.shape)
print(marks.head()); print(marks.shape)
students.info()
print(students.isna().sum())
print(marks.isna().sum())
print("Student_ID unique:", students["Student_ID"].is_unique)
```

## Q6. Stack two sales files

```python
print(pd.concat([jan, feb]))
print(pd.concat([jan, feb], ignore_index=True))
```

**Answer:** Without `ignore_index`, the original row numbers are kept (0,1,2,0,1,2), so the index repeats. With `ignore_index=True`, the rows are renumbered 0 to 5.

## Q7. Three months and total units

```python
all_sales = pd.concat([jan, feb, mar], ignore_index=True)
print(all_sales)
print(all_sales.groupby("Product")["Units"].sum())
```

## Q8. Place tables side by side

```python
att = attendance.set_index("Student_ID")
res = results.set_index("Student_ID")
side = pd.concat([att, res], axis=1)
print(side)
print(side.shape)
```

## Q9. Merge with an inner join

```python
inner = pd.merge(students, marks, on="Student_ID", how="inner")
print(inner)
print(len(inner))
```

## Q10. Left join and missing marks

```python
left = pd.merge(students, marks, on="Student_ID", how="left")
print(left)
print(left.isna().sum())
print(left.loc[left["Data_Science_Marks"].isna(), "Name"].tolist())
```

## Q11. Compare the join types

```python
for how in ["inner", "left", "right", "outer"]:
    print(how, len(pd.merge(students, marks, on="Student_ID", how=how)))
outer = pd.merge(students, marks, on="Student_ID", how="outer", indicator=True)
print(outer)
print("Not in both tables:", outer.loc[outer["_merge"] != "both", "Student_ID"].tolist())
```

**Answer:** inner = 5, left = 8, right = 6, outer = 9. IDs not in both tables: 103, 107, 108 (only in `students`) and 109 (only in `marks`).

## Q12. Join using the index

```python
student_info = students.set_index("Student_ID")
student_marks = marks.set_index("Student_ID")
joined = student_info.join(student_marks, how="left")
print(joined)
print(joined.reset_index().equals(left))
```

**Answer:** It gives the same rows and values as the left join in Q10. The only difference is that `Student_ID` is the index here instead of a normal column.

## Q13. Long to wide with pivot

```python
sales = pd.concat([jan, feb], ignore_index=True)
print(sales.pivot(index="Product", columns="Month", values="Units"))
```

## Q14. When pivot fails

```python
try:
    transactions.pivot(index="Product", columns="Month", values="Units")
except ValueError as e:
    print("Error:", e)

pt = transactions.pivot_table(index="Product", columns="Month", values="Units", aggfunc="sum")
print(pt)
```

**Why pivot() failed:** Some Product-Month pairs (like Pen in Jan) occur more than once, and `pivot()` cannot decide which value to put in one cell. `pivot_table` solves this by aggregating the duplicates.

## Q15. Missing combinations in pivot_table

```python
pt = transactions.pivot_table(index="Product", columns="Month", values="Units", aggfunc="sum")
print(pt)   # Pencil / Feb is NaN
print(transactions.pivot_table(index="Product", columns="Month", values="Units", aggfunc="sum", fill_value=0))
print(transactions.pivot_table(index="Product", columns="Month", values="Units", aggfunc="mean"))
```

## Q21. Merge three tables

```python
analysis_df = pd.merge(students, attendance, on="Student_ID", how="left")
analysis_df = pd.merge(analysis_df, results, on="Student_ID", how="left")
print(analysis_df)
print(analysis_df.shape)
```

## Q22. Department summary

```python
summary = analysis_df.groupby("Department")[["Attendance_Percent", "Final_Marks"]].mean().round(1)
print(summary)
```

## Q23. Attendance groups

```python
analysis_df["Attendance_Group"] = np.where(analysis_df["Attendance_Percent"] >= 85, "High", "Low")
print(analysis_df.pivot_table(index="Department", columns="Attendance_Group", values="Final_Marks", aggfunc="mean"))
```

## Q24. Safe merge with validate

```python
bad_marks = pd.concat([marks, marks.iloc[[1]]], ignore_index=True)
bad_merge = pd.merge(students, bad_marks, on="Student_ID", how="left")
print(len(bad_merge), "rows (original students had 8)")
print(bad_merge)
try:
    pd.merge(students, bad_marks, on="Student_ID", how="left", validate="one_to_one")
except Exception as e:
    print("Error:", e)
```

**Answer:** The merge silently gives 9 rows instead of 8 because student 102 is duplicated. With `validate="one_to_one"` pandas raises a `MergeError` and stops, so the problem is caught.
